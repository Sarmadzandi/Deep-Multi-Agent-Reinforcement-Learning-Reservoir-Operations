```python
class LandUseModel:
    def __init__(self):
        #----------
        self.Zone_Types = ["Zone-5", "Zone-9", "Zone-18", "Zone-21", "Zone-22", "Zone-Es"] # -> Constant
        self.Space_Types = ["UrbanSpace", "WaterSpace", "AgricSpace", "GreenSpace", "WasteSpace"] # -> Constant
        self.Tree_Types = ['kajteh', 'aghaghia', 'sarvenoghreyi', 'chenar', 'sarvekhoreyi', 'tootsefid', 'zabangonjeshk', 'sarvenaz', 'narvan', 'chaman'] # -> Constant
        #----------
        self.BaselineYear, self.FinalYear = 2013, 2040 # -> Constant
        #----------
        self.Areas = {year: {zone: {space: 0 for space in self.Space_Types} for zone in self.Zone_Types} for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        self.Areas[2013] = {'Zone-5': {'UrbanSpace': 2427.75, 'WaterSpace': 34.85, 'AgricSpace': 52.52, 'GreenSpace': 787.41, 'WasteSpace': 364.52},
                            'Zone-9': {'UrbanSpace': 850.10, 'WaterSpace': 20.99, 'AgricSpace': 0.00, 'GreenSpace': 295.02, 'WasteSpace': 521.91},
                            'Zone-18': {'UrbanSpace': 2148.63, 'WaterSpace': 10.77, 'AgricSpace': 272.30, 'GreenSpace': 822.23, 'WasteSpace': 605.23},
                            'Zone-21': {'UrbanSpace': 1911.36, 'WaterSpace': 27.15, 'AgricSpace': 0.00, 'GreenSpace': 392.99, 'WasteSpace': 887.01},
                            'Zone-22': {'UrbanSpace': 800.28, 'WaterSpace': 142.68, 'AgricSpace': 0.00, 'GreenSpace': 843.45, 'WasteSpace': 951.02},
                            'Zone-Es': {'UrbanSpace': 5367.08, 'WaterSpace': 30.95, 'AgricSpace': 17205.31, 'GreenSpace': 450.59, 'WasteSpace': 7183.15}
                            } # -> Initialization
        #----------
        self.Scenarios = {'Scenario1': {'Zone-5': {'UrbanSpace': 2738.33, 'WaterSpace': 5.53, 'AgricSpace': 52.52, 'GreenSpace': 683.96, 'WasteSpace': 187.27},
                                        'Zone-9': {'UrbanSpace': 945.08, 'WaterSpace': 0.00, 'AgricSpace': 0.00, 'GreenSpace': 152.17, 'WasteSpace': 591.38},
                                        'Zone-18': {'UrbanSpace': 2657.89, 'WaterSpace': 0.12, 'AgricSpace': 153.08, 'GreenSpace': 612.33, 'WasteSpace': 437.05},
                                        'Zone-21': {'UrbanSpace': 2441.98, 'WaterSpace': 0.00, 'AgricSpace': 0.00, 'GreenSpace': 201.08, 'WasteSpace': 575.55},
                                        'Zone-22': {'UrbanSpace': 1309.25, 'WaterSpace': 141.93, 'AgricSpace': 0.00, 'GreenSpace': 734.32, 'WasteSpace': 551.42},
                                        'Zone-Es': {'UrbanSpace': 8253.92, 'WaterSpace': 0.00, 'AgricSpace': 14475.81, 'GreenSpace': 224.82, 'WasteSpace': 7284.88}
                                        },
                          'Scenario2': {'Zone-5': {'UrbanSpace': 2785.14, 'WaterSpace': 5.58, 'AgricSpace': 52.52, 'GreenSpace': 646.06, 'WasteSpace': 179.19},
                                        'Zone-9': {'UrbanSpace': 981.09, 'WaterSpace': 0.00, 'AgricSpace': 0.00, 'GreenSpace': 118.44, 'WasteSpace': 589.77},
                                        'Zone-18': {'UrbanSpace': 2765.07, 'WaterSpace': 0.18, 'AgricSpace': 132.44, 'GreenSpace': 538.16, 'WasteSpace': 424.98},
                                        'Zone-21': {'UrbanSpace': 2511.81, 'WaterSpace': 0.00, 'AgricSpace': 0.00, 'GreenSpace': 137.70, 'WasteSpace': 570.06},
                                        'Zone-22': {'UrbanSpace': 1467.36, 'WaterSpace': 141.30, 'AgricSpace': 0.00, 'GreenSpace': 609.30, 'WasteSpace': 521.37},
                                        'Zone-Es': {'UrbanSpace': 10310.13, 'WaterSpace': 0.00, 'AgricSpace': 12915.00, 'GreenSpace': 122.76, 'WasteSpace': 6906.96}
                                        },
                          'Scenario3': {'Zone-5': {'UrbanSpace': 2785.14, 'WaterSpace': 5.58, 'AgricSpace': 52.52, 'GreenSpace': 804.46, 'WasteSpace': 20.79},
                                        'Zone-9': {'UrbanSpace': 981.09, 'WaterSpace': 0.00, 'AgricSpace': 0.00, 'GreenSpace': 400.32, 'WasteSpace': 307.89},
                                        'Zone-18': {'UrbanSpace': 2765.07, 'WaterSpace': 0.18, 'AgricSpace': 132.44, 'GreenSpace': 850.82, 'WasteSpace': 112.32},
                                        'Zone-21': {'UrbanSpace': 2511.81, 'WaterSpace': 0.00, 'AgricSpace': 0.00, 'GreenSpace': 560.97, 'WasteSpace': 146.79},
                                        'Zone-22': {'UrbanSpace': 1467.36, 'WaterSpace': 141.30, 'AgricSpace': 0.00, 'GreenSpace': 1092.69, 'WasteSpace': 37.98},
                                        'Zone-Es': {'UrbanSpace': 10310.13, 'WaterSpace': 0.00, 'AgricSpace': 12915.54, 'GreenSpace': 311.85, 'WasteSpace': 6717.33}
                                        },
                          'Scenario4': {'Zone-5': {'UrbanSpace': 2247.84, 'WaterSpace': 5.58, 'AgricSpace': 52.52, 'GreenSpace': 1216.66, 'WasteSpace': 145.89},
                                        'Zone-9': {'UrbanSpace': 829.08, 'WaterSpace': 0.00, 'AgricSpace': 0.00, 'GreenSpace': 278.01, 'WasteSpace': 582.21},
                                        'Zone-18': {'UrbanSpace': 2384.19, 'WaterSpace': 0.18, 'AgricSpace': 156.42, 'GreenSpace': 948.60, 'WasteSpace': 371.43},
                                        'Zone-21': {'UrbanSpace': 2297.43, 'WaterSpace': 0.00, 'AgricSpace': 0.00, 'GreenSpace': 396.72, 'WasteSpace': 525.42},
                                        'Zone-22': {'UrbanSpace': 1068.30, 'WaterSpace': 141.30, 'AgricSpace': 0.00, 'GreenSpace': 1130.94, 'WasteSpace': 398.79},
                                        'Zone-Es': {'UrbanSpace': 8133.39, 'WaterSpace': 0.00, 'AgricSpace': 14439.60, 'GreenSpace': 414.00, 'WasteSpace': 7267.86}
                                        }
                          } # -> Initialization
        #----------
        self.Carbon_Pools = {'UrbanSpace': 25.0, 'WaterSpace': 5.0, 'AgricSpace': 72.6, 'GreenSpace': 24.8, 'WasteSpace': 61.2} # -> Constant
        self.Carbon_Storage = {year: {zone: {space: 0 for space in self.Space_Types} for zone in self.Zone_Types} for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        self.Total_Carbon_Storage = {year: 0 for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        #----------
        self.Max_Temperature = {2013: 23.81, 2014: 23.39, 2015: 23.69, 2016: 23.94, 2017: 23.74, 2018: 23.98, 2019: 23.33,
                                2020: 23.33, 2021: 24.67, 2022: 24.43, 2023: 23.87, 2024: 24.49, 2025: 24.26, 2026: 24.48,
                                2027: 24.14, 2028: 24.57, 2029: 24.49, 2030: 24.36, 2031: 24.29, 2032: 24.41, 2033: 24.43,
                                2034: 24.75, 2035: 24.74, 2036: 24.54, 2037: 24.65, 2038: 24.50, 2039: 24.15, 2040: 24.73} # -> Constant

        self.Mean_Temperature = {2013: 35.95, 2014: 36.43, 2015: 36.83, 2016: 35.77, 2017: 35.97, 2018: 36.95, 2019: 36.79,
                                 2020: 36.79, 2021: 37.00, 2022: 36.95, 2023: 36.11, 2024: 36.52, 2025: 36.64, 2026: 36.93,
                                 2027: 36.31, 2028: 36.88, 2029: 36.80, 2030: 36.25, 2031: 36.52, 2032: 36.67, 2033: 36.38,
                                 2034: 36.66, 2035: 36.91, 2036: 36.75, 2037: 37.03, 2038: 36.84, 2039: 36.74, 2040: 36.83} # -> Constant
        #----------
        self.Crop_Types = ['Grape', 'Apple', 'Wheat', 'Barley', 'Alfalfa'] # -> Constant
        self.Crop_Production = {year: {zone: {space: 0 for space in self.Space_Types} for zone in self.Zone_Types} for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        self.Total_Crop_Production = {year: 0 for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        #----------
        self.Biogas_Crop = {'Grape': 100.51, 'Apple': 228, 'Wheat': 270, 'Barley': 300, 'Alfalfa': 420} # -> Constant
        self.Biobutanol_Crop = {'Grape': 34.72, 'Apple': 51.69, 'Wheat': 98.35, 'Barley': 96.81, 'Alfalfa': 88.27} # -> Constant
        self.Bioethanol_Crop = {'Grape': 70, 'Apple': 65, 'Wheat': 432, 'Barley': 350, 'Alfalfa': 451} # -> Constant
        self.Volume_to_Energy = {'Biogas': 6, 'Biobutanol': 7.83, 'Bioethanol': 5.9} # -> Constant
        self.Residual_Percent = {'Grape': 0.5, 'Apple': 0.5, 'Wheat': 1.75, 'Barley': 1.2, 'Alfalfa': 0.15} # -> Constant
        self.Percent_of_Total_Crop = {'Grape': 0.1, 'Apple': 0.3, 'Wheat': 0.2, 'Barley': 0.13, 'Alfalfa': 0.2} # -> Constant
        self.Bioenergy = {year: {zone: {'Biogas': 0, 'Biobutanol': 0, 'Bioethanol': 0} for zone in ['Zone-5', 'Zone-18', 'Zone-Es']} for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        self.Total_Bioenergy = {year: {zone: 0 for zone in ['Zone-5', 'Zone-18', 'Zone-Es']} for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        #----------
        self.Urban_Population = {year: {zone: 0 for zone in self.Zone_Types} for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        self.Urban_Population[2013] = {'Zone-5': 565584, 'Zone-9': 141499, 'Zone-18': 402774, 'Zone-21': 107812, 'Zone-22': 69392, 'Zone-Es': 275235} # -> Initialization
        self.Urban_Population[2040] = {'Zone-5': 700044, 'Zone-9': 171832, 'Zone-18': 513060, 'Zone-21': 158295, 'Zone-22': 160511, 'Zone-Es': 374262} # -> Initialization

        self.Urban_Rural_Population = {year: {zone: 0 for zone in self.Zone_Types} for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        self.Urban_Rural_Population[2013] = {'Zone-5': 565584, 'Zone-9': 141499, 'Zone-18': 402774, 'Zone-21': 107812, 'Zone-22': 69392, 'Zone-Es': 328061} # -> Initialization
        self.Urban_Rural_Population[2040] = {'Zone-5': 700044, 'Zone-9': 171832, 'Zone-18': 513060, 'Zone-21': 158295, 'Zone-22': 160511, 'Zone-Es': 454806} # -> Initialization

        self.Total_Urban_Population = {year: 0 for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        self.Total_Urban_Population[2013] = 1562296 # -> Initialization
        self.Total_Urban_Population[2040] = 2078004 # -> Initialization

        self.Total_Urban_Rural_Population = {year: 0 for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        self.Total_Urban_Rural_Population[2013] = 1615122 # -> Initialization
        self.Total_Urban_Rural_Population[2040] = 2158548 # -> Initialization
        #----------
        self.GreenSpace_per_UrbanPopulation = {year: {zone: 0 for zone in self.Zone_Types} for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        self.Total_GreenSpace_per_UrbanPopulation = {year: 0 for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        #----------
        self.Heat_Island_Effect = {'Zone-5': 4, 'Zone-9': 4.5, 'Zone-18': 3, 'Zone-21': 4.5, 'Zone-22': 4, 'Zone-Es': 2.5} # -> Constant

        self.Heat_Mitigation_Index = {scenario: {year: {zone: 0 for zone in self.Zone_Types} for year in range(self.BaselineYear, self.FinalYear + 1)} for scenario in ['Scenario1','Scenario2','Scenario3','Scenario4']} # -> Variable
        self.Heat_Mitigation_Index['Scenario1'][2013] = {'Zone-5': 0.32, 'Zone-9': 0.27, 'Zone-18': 0.34, 'Zone-21': 0.23, 'Zone-22': 0.38, 'Zone-Es': 0.53} # -> Initialization
        self.Heat_Mitigation_Index['Scenario1'][2040] = {'Zone-5': 0.31, 'Zone-9': 0.22, 'Zone-18': 0.29, 'Zone-21': 0.22, 'Zone-22': 0.36, 'Zone-Es': 0.47} # -> Initialization

        self.Heat_Mitigation_Index['Scenario2'][2013] = {'Zone-5': 0.32, 'Zone-9': 0.27, 'Zone-18': 0.34, 'Zone-21': 0.23, 'Zone-22': 0.38, 'Zone-Es': 0.53} # -> Initialization
        self.Heat_Mitigation_Index['Scenario2'][2040] = {'Zone-5': 0.31, 'Zone-9': 0.22, 'Zone-18': 0.28, 'Zone-21': 0.21, 'Zone-22': 0.34, 'Zone-Es': 0.45} # -> Initialization

        self.Heat_Mitigation_Index['Scenario3'][2013] = {'Zone-5': 0.32, 'Zone-9': 0.27, 'Zone-18': 0.34, 'Zone-21': 0.23, 'Zone-22': 0.38, 'Zone-Es': 0.53} # -> Initialization
        self.Heat_Mitigation_Index['Scenario3'][2040] = {'Zone-5': 0.33, 'Zone-9': 0.32, 'Zone-18': 0.34, 'Zone-21': 0.29, 'Zone-22': 0.46, 'Zone-Es': 0.45} # -> Initialization

        self.Heat_Mitigation_Index['Scenario4'][2013] = {'Zone-5': 0.32, 'Zone-9': 0.27, 'Zone-18': 0.34, 'Zone-21': 0.23, 'Zone-22': 0.38, 'Zone-Es': 0.53} # -> Initialization
        self.Heat_Mitigation_Index['Scenario4'][2040] = {'Zone-5': 0.42, 'Zone-9': 0.28, 'Zone-18': 0.38, 'Zone-21': 0.26, 'Zone-22': 0.51, 'Zone-Es': 0.47} # -> Initialization

        self.Temperature_Difference = 1.38 # -> Constant
        self.Zone_Temperature = {year: {zone: 0 for zone in self.Zone_Types} for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        #----------
        self.Consuming_Water_per_Person = {
            2013: {'Zone-5': 316, 'Zone-9': 316, 'Zone-18': 316, 'Zone-21': 316, 'Zone-22': 316, 'Zone-Es': 235},
            2014: {'Zone-5': 316, 'Zone-9': 316, 'Zone-18': 316, 'Zone-21': 316, 'Zone-22': 316, 'Zone-Es': 235},
            2015: {'Zone-5': 316, 'Zone-9': 316, 'Zone-18': 316, 'Zone-21': 316, 'Zone-22': 316, 'Zone-Es': 235},
            2016: {'Zone-5': 316, 'Zone-9': 316, 'Zone-18': 316, 'Zone-21': 316, 'Zone-22': 316, 'Zone-Es': 235},
            2017: {'Zone-5': 314, 'Zone-9': 314, 'Zone-18': 314, 'Zone-21': 314, 'Zone-22': 314, 'Zone-Es': 235},
            2018: {'Zone-5': 312, 'Zone-9': 312, 'Zone-18': 312, 'Zone-21': 312, 'Zone-22': 312, 'Zone-Es': 235},
            2019: {'Zone-5': 310, 'Zone-9': 310, 'Zone-18': 310, 'Zone-21': 310, 'Zone-22': 310, 'Zone-Es': 235},
            2020: {'Zone-5': 308, 'Zone-9': 308, 'Zone-18': 308, 'Zone-21': 308, 'Zone-22': 308, 'Zone-Es': 235},
            2021: {'Zone-5': 307, 'Zone-9': 307, 'Zone-18': 307, 'Zone-21': 307, 'Zone-22': 307, 'Zone-Es': 235},
            2022: {'Zone-5': 305, 'Zone-9': 305, 'Zone-18': 305, 'Zone-21': 305, 'Zone-22': 305, 'Zone-Es': 235},
            2023: {'Zone-5': 303, 'Zone-9': 303, 'Zone-18': 303, 'Zone-21': 303, 'Zone-22': 303, 'Zone-Es': 235},
            2024: {'Zone-5': 301, 'Zone-9': 301, 'Zone-18': 301, 'Zone-21': 301, 'Zone-22': 301, 'Zone-Es': 235},
            2025: {'Zone-5': 299, 'Zone-9': 299, 'Zone-18': 299, 'Zone-21': 299, 'Zone-22': 299, 'Zone-Es': 235},
            2026: {'Zone-5': 297, 'Zone-9': 297, 'Zone-18': 297, 'Zone-21': 297, 'Zone-22': 297, 'Zone-Es': 235},
            2027: {'Zone-5': 296, 'Zone-9': 296, 'Zone-18': 296, 'Zone-21': 296, 'Zone-22': 296, 'Zone-Es': 235},
            2028: {'Zone-5': 295, 'Zone-9': 295, 'Zone-18': 295, 'Zone-21': 295, 'Zone-22': 295, 'Zone-Es': 235},
            2029: {'Zone-5': 294, 'Zone-9': 294, 'Zone-18': 294, 'Zone-21': 294, 'Zone-22': 294, 'Zone-Es': 235},
            2030: {'Zone-5': 293, 'Zone-9': 293, 'Zone-18': 293, 'Zone-21': 293, 'Zone-22': 293, 'Zone-Es': 235},
            2031: {'Zone-5': 293, 'Zone-9': 293, 'Zone-18': 293, 'Zone-21': 293, 'Zone-22': 293, 'Zone-Es': 235},
            2032: {'Zone-5': 292, 'Zone-9': 292, 'Zone-18': 292, 'Zone-21': 292, 'Zone-22': 292, 'Zone-Es': 235},
            2033: {'Zone-5': 291, 'Zone-9': 291, 'Zone-18': 291, 'Zone-21': 291, 'Zone-22': 291, 'Zone-Es': 235},
            2034: {'Zone-5': 290, 'Zone-9': 290, 'Zone-18': 290, 'Zone-21': 290, 'Zone-22': 290, 'Zone-Es': 235},
            2035: {'Zone-5': 289, 'Zone-9': 289, 'Zone-18': 289, 'Zone-21': 289, 'Zone-22': 289, 'Zone-Es': 235},
            2036: {'Zone-5': 288, 'Zone-9': 288, 'Zone-18': 288, 'Zone-21': 288, 'Zone-22': 288, 'Zone-Es': 235},
            2037: {'Zone-5': 288, 'Zone-9': 288, 'Zone-18': 288, 'Zone-21': 288, 'Zone-22': 288, 'Zone-Es': 235},
            2038: {'Zone-5': 288, 'Zone-9': 288, 'Zone-18': 288, 'Zone-21': 288, 'Zone-22': 288, 'Zone-Es': 235},
            2039: {'Zone-5': 288, 'Zone-9': 288, 'Zone-18': 288, 'Zone-21': 288, 'Zone-22': 288, 'Zone-Es': 235},
            2040: {'Zone-5': 288, 'Zone-9': 288, 'Zone-18': 288, 'Zone-21': 288, 'Zone-22': 288, 'Zone-Es': 235}
        } # -> Constant

        self.Total_Consuming_Water = {year: {zone: 0 for zone in self.Zone_Types} for year in range(self.BaselineYear, self.FinalYear + 1)} # -> Variable
        #----------
        self.Rainfall = {2013: 154.15, 2014: 116.52, 2015: 204.53, 2016: 167.80, 2017: 166.36, 2018: 259.14, 2019: 334.57,
                         2020: 194.90, 2021: 153.70, 2022: 188.00, 2023: 304.80, 2024: 229.80, 2025: 195.80, 2026: 189.70,
                         2027: 136.50, 2028: 173.60, 2029: 257.90, 2030: 143.40, 2031: 224.10, 2032: 190.10, 2033: 199.70,
                         2034: 183.80, 2035: 202.50, 2036: 243.90, 2037: 251.00, 2038: 189.60, 2039: 245.20, 2040: 194.70
                         }
        self.Evapotranspiration = {2013: 573.57, 2014: 550.15, 2015: 553.34, 2016: 570.88, 2017: 554.73, 2018: 543.30, 2019: 549.08,
                                   2020: 588.58, 2021: 566.12, 2022: 551.21, 2023: 538.09, 2024: 557.14, 2025: 551.37, 2026: 557.76,
                                   2027: 551.59, 2028: 561.13, 2029: 553.22, 2030: 554.88, 2031: 559.53, 2032: 558.81, 2033: 545.87,
                                   2034: 565.63, 2035: 567.57, 2036: 559.85, 2037: 563.95, 2038: 560.19, 2039: 551.16, 2040: 571.13
                                   }
        self.Density_factor = 1
        self.Microclimate_factor = 1
        self.Tree_Species = {'kajteh': 0.3, 'aghaghia': 0.2, 'sarvenoghreyi': 0.3, 'chenar': 0.6, 'sarvekhoreyi': 0.3, 'tootsefid': 0.3, 'zabangonjeshk': 0.3, 'sarvenaz': 0.6, 'narvan': 0.4, 'chaman': 0.9}
        self.Tree_Species_coefficients = {'kajteh': 0.13,'aghaghia': 0.1, 'sarvenoghreyi': 0.094, 'chenar': 0.084,'sarvekhoreyi': 0.072, 'tootsefid': 0.057, 'zabangonjeshk': 0.047,'sarvenaz': 0.035, 'narvan': 0.035, 'chaman': 0.35}
        self.Tree_Coefficient = {tree: 0 for tree in self.Tree_Types}
        self.GreenSpace_Water_Demand = {year: {zone: 0 for zone in self.Zone_Types} for year in range(self.BaselineYear, self.FinalYear + 1)}
        self.Total_GreenSpace_Water_Demand = {year: 0 for year in range(self.BaselineYear, self.FinalYear + 1)}
        #----------

    def linear_interpolation(self, year, start_value, end_value):
        return start_value + ((year - self.BaselineYear) * (end_value - start_value) / (self.FinalYear - self.BaselineYear))

    def calculate_area(self, scenario, year, zone, space):
        Area2013 = self.Areas[2013][zone][space]
        Area2040 = self.Scenarios[scenario][zone][space]
        self.Areas[year][zone][space] = self.linear_interpolation(year, Area2013, Area2040)
        return self.Areas[year][zone][space]

    def calculate_carbon_storage(self, scenario, year):
        for zone in self.Zone_Types:
            for space in self.Space_Types:
                self.calculate_area(scenario, year, zone, space)
                self.Carbon_Storage[year][zone][space] = self.Areas[year][zone][space] * self.Carbon_Pools[space]
                self.Total_Carbon_Storage[year] += self.Carbon_Storage[year][zone][space]

        return self.Carbon_Storage[year], self.Total_Carbon_Storage[year]

    def calculate_crop_yield(self, year):
        max_temp = self.Max_Temperature[year]
        crop_yield = 59.271 - 1.785 * max_temp
        return max(crop_yield, 0)  # Ensure non-negative yield

    def calculate_crop_production(self, scenario, year):
        for zone in self.Zone_Types:
          self.calculate_area(scenario, year, zone,'AgricSpace')
          yield_per_hectare = self.calculate_crop_yield(year)
          self.Crop_Production[year][zone]['AgricSpace'] = self.Areas[year][zone]['AgricSpace'] * yield_per_hectare
          self.Total_Crop_Production[year] += self.Crop_Production[year][zone]['AgricSpace']

        return self.Crop_Production[year], self.Total_Crop_Production[year]

    def calculate_bioenergy_potential(self, scenario, year):
        self.calculate_crop_production(scenario, year)

        for zone in ['Zone-5', 'Zone-18', 'Zone-Es']:
            if zone in ['Zone-5', 'Zone-18']:
                crops = ['Grape', 'Apple']
            else:  # Zone-Es
                crops = ['Wheat', 'Barley', 'Alfalfa']

            for energy_type in ['Biogas', 'Biobutanol', 'Bioethanol']:
                energy_crop = getattr(self, f"{energy_type}_Crop")
                self.Bioenergy[year][zone][energy_type] = (
                    self.Crop_Production[year][zone]['AgricSpace'] *
                    sum(energy_crop[crop] * 0.35 * self.Percent_of_Total_Crop[crop] * self.Residual_Percent[crop]
                        for crop in crops)
                ) * self.Volume_to_Energy[energy_type]

            self.Total_Bioenergy[year][zone] = sum(self.Bioenergy[year][zone].values()) * 0.42

        return self.Bioenergy[year], self.Total_Bioenergy[year]

    def calculate_urban_rural_population(self, year):
        for zone in self.Zone_Types:
            # Urban
            pop2013 = self.Urban_Population[2013][zone]
            pop2040 = self.Urban_Population[2040][zone]
            self.Urban_Population[year][zone] = self.linear_interpolation(year, pop2013, pop2040)
            self.Total_Urban_Population[year] += self.Urban_Population[year][zone]
            # Urban + Rural
            Pop2013 = self.Urban_Rural_Population[2013][zone]
            Pop2040 = self.Urban_Rural_Population[2040][zone]
            self.Urban_Rural_Population[year][zone] = self.linear_interpolation(year, Pop2013, Pop2040)
            self.Total_Urban_Rural_Population[year] += self.Urban_Rural_Population[year][zone]

        return self.Urban_Population[year], self.Total_Urban_Population[year], self.Urban_Rural_Population[year], self.Total_Urban_Rural_Population[year]

    def calculate_green_space_per_population(self, scenario, year):
        self.calculate_urban_rural_population(year)
        for zone in self.Zone_Types:
            self.GreenSpace_per_UrbanPopulation[year][zone] = (self.Areas[year][zone]['GreenSpace'] / self.Urban_Population[year][zone]) * 10000
            self.Total_GreenSpace_per_UrbanPopulation[year] += self.GreenSpace_per_UrbanPopulation[year][zone]

        return self.GreenSpace_per_UrbanPopulation[year], self.Total_GreenSpace_per_UrbanPopulation[year]

    def calculate_zone_temperature(self, scenario, year):
        for zone in self.Zone_Types:
            heat_mitigation_index_2013 = self.Heat_Mitigation_Index[scenario][2013][zone]
            heat_mitigation_index_2040 = self.Heat_Mitigation_Index[scenario][2040][zone]
            self.Heat_Mitigation_Index[scenario][year][zone] = self.linear_interpolation(year, heat_mitigation_index_2013, heat_mitigation_index_2040)
            self.Zone_Temperature[year][zone] = (self.Mean_Temperature[year] - self.Temperature_Difference) + ((1 - self.Heat_Mitigation_Index[scenario][year][zone]) * self.Heat_Island_Effect[zone])

        return self.Zone_Temperature[year]

    def calculate_consuming_water(self, year):
        self.calculate_urban_rural_population(year)
        for zone in self.Zone_Types:
          self.Total_Consuming_Water[year][zone] = self.Consuming_Water_per_Person[year][zone] * 0.001 * 0.000001 * 365 * self.Urban_Rural_Population[year][zone]

        return self.Total_Consuming_Water[year]

    def green_space_water(self, year):
        for tree in self.Tree_Types:
            self.Tree_Coefficient[tree] = self.Tree_Species[tree] * self.Density_factor * self.Microclimate_factor

        Effective_Rainfall = 0.91 * self.Rainfall[year] - 55.1

        for zone in self.Zone_Types:
            self.GreenSpace_Water_Demand[year][zone] = sum(
                ((self.Tree_Coefficient[tree] * self.Evapotranspiration[year] - Effective_Rainfall) / 0.35) * self.Tree_Species_coefficients[tree]
                for tree in self.Tree_Types
            ) * self.Area[year][zone]['GreenSpace'] * 0.00001

            self.Total_GreenSpace_Water_Demand[year] += self.GreenSpace_Water_Demand[year][zone]
        return self.GreenSpace_Water_Demand[year], self.Total_GreenSpace_Water_Demand[year]
```

```python
def main():
    # Initialize the model
    model = LandUseModel()

    # Define scenarios and years to analyze
    scenarios = ['Scenario1', 'Scenario2', 'Scenario3', 'Scenario4']
    years = [2013, 2025, 2040]

    for scenario in scenarios:
        print(f"\nAnalyzing {scenario}:")
        for year in years:
            print(f"\nYear {year}:")

            # Calculate and display carbon storage
            carbon_storage, total_carbon_storage = model.calculate_carbon_storage(scenario, year)
            print(f"Total Carbon Storage: {total_carbon_storage:.2f}")

            # Calculate and display crop production
            crop_production, total_crop_production = model.calculate_crop_production(scenario, year)
            print(f"Total Crop Production: {total_crop_production:.2f}")

            # Calculate and display bioenergy potential
            bioenergy, total_bioenergy = model.calculate_bioenergy_potential(scenario, year)
            print(f"Total Bioenergy Potential: {sum(total_bioenergy.values()):.2f}")

            # Calculate and display urban and rural population
            urban_pop, total_urban_pop, urban_rural_pop, total_urban_rural_pop = model.calculate_urban_rural_population(year)
            print(f"Total Urban Population: {total_urban_pop:.0f}")
            print(f"Total Urban+Rural Population: {total_urban_rural_pop:.0f}")

            # Calculate and display green space per population
            green_space_per_pop, total_green_space_per_pop = model.calculate_green_space_per_population(scenario, year)
            print(f"Average Green Space per Urban Population: {total_green_space_per_pop / len(model.Zone_Types):.2f} m²/person")

            # Calculate and display zone temperatures
            zone_temperatures = model.calculate_zone_temperature(scenario, year)
            avg_temp = sum(zone_temperatures.values()) / len(zone_temperatures)
            print(f"Average Zone Temperature: {avg_temp:.2f}°C")

    # Example of accessing specific data
    print("\nDetailed data for Zone-5 in Scenario1, year 2025:")
    print(f"Carbon Storage: {model.Carbon_Storage[2025]['Zone-5']}")
    print(f"Crop Production: {model.Crop_Production[2025]['Zone-5']}")
    print(f"Bioenergy Potential: {model.Bioenergy[2025]['Zone-5']}")
    print(f"Urban Population: {model.Urban_Population[2025]['Zone-5']}")
    print(f"Green Space per Population: {model.GreenSpace_per_UrbanPopulation[2025]['Zone-5']}")
    print(f"Zone Temperature: {model.Zone_Temperature[2025]['Zone-5']}")

if __name__ == "__main__":
    main()
```
